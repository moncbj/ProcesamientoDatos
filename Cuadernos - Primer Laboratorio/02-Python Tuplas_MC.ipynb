## **Introducción**

<p><strong>¡Bienvenido!</strong> En este cuaderno aprenderás sobre las tuplas en el Lenguaje de Programación Python. Al final de este laboratorio, conocerás lo básico sobre las operaciones con tuplas en Python, esto incluye la indexación, el slicing y la clasificación de las tuplas.</p>

En este cuaderno se trabaja con las tuplas en Python, un tipo de dato que permite almacenar una colección ordenada e inmutable de elementos. Se abordan operaciones como la indexación positiva y negativa, la concatenación, el slicing, el ordenamiento y el anidamiento de tuplas.


## Objetivo General

Comprender y aplicar las operaciones fundamentales con tuplas en Python para el almacenamiento y manipulación de datos ordenados e inmutables.

## Objetivos Específicos

- Aplicar indexación positiva y negativa, así como slicing, para acceder a los elementos de una tupla.
- Concatenar tuplas y ordenar sus elementos utilizando la función `sorted()`.
- Acceder a elementos dentro de tuplas anidadas mediante múltiples niveles de indexación.

## Resumen de Conceptos Clave

**Tupla:** Colección ordenada e inmutable de elementos, que puede almacenar distintos tipos de datos (cadenas, enteros, decimales, e incluso otras tuplas).

**Indexación:** Permite acceder a un elemento específico de la tupla mediante su posición (positiva desde el inicio, negativa desde el final).

**Concatenación:** Se realiza con el operador `+`, y permite combinar dos o más tuplas en una nueva.

**Slicing:** Permite extraer un subconjunto de elementos de la tupla a partir de un rango de índices.

**Ordenamiento:** La función `sorted()` devuelve una nueva lista con los elementos de la tupla ordenados, sin modificar la tupla original (ya que es inmutable).

**Anidamiento:** Una tupla puede contener otras tuplas, a las que se accede usando índices sucesivos, por ejemplo `NestedT[2][1][0]`.

<h2 id="tuple">Tuplas</h2>

En Python existen diferentes tipos de datos: cadenas, números enteros o decimales (float), etc. Todos estos tipos de datos pueden ser almacenados en una tupla como se indica a continuación:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/TuplesType.png" width="750" align="center" />

Vamos a crear tu primera tupla compuesta de: una cadena, un entero y un número decimal.

```python
# Crear tu primera tupla

tuple1=(1,2,3)
tuple1

pelicula1 = ("El Padrino", 1972, 175, "Drama", 245)
pelicula1
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crean dos tuplas usando paréntesis: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">tuple1=(1,2,3)</code>, con tres números enteros, y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">pelicula1 = (&quot;El Padrino&quot;, 1972, 175, &quot;Drama&quot;, 245)</code>, que combina texto, enteros y datos numéricos en una sola colección ordenada.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Como no se usa <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">print()</code>, Jupyter solo muestra el resultado de la última línea de la celda; por eso solo se ve el contenido de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">pelicula1</code>.</p>
</div>

El tipo de dato almacenado en la variable es **tuple**.

```python
# Imprime el tipo de dato de la tupla

type(tuple1)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">type(tuple1)</code> devuelve el tipo de dato de la variable, que en este caso es <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">tuple</code>, confirmando que se trata de una tupla.</p>
</div>

<h3 id="index">Indexación</h3>

Cada elemento dentro de la tupla puede ser accedido a través de su número de índice. La siguiente tabla representa la relación entre el número de índice y los elementos dentro de la tupla. Podemos imprimir cada uno de los valores de la tupla:

```python
# Imprimir la variable en cada índice

print(tuple1[0])
print(tuple1[1])
print(tuple1[2])

print(pelicula1[0])
```

Podemos concatenar o combinar tuplas utilizando el signo **+**:

```python
# Concatenar dos tuplas

tuple2 = tuple1 + ("hard rock", 10)
tuple2

pelicula2 = pelicula1 + ("Francis Ford Coppola",)
pelicula2
```

Podemos aplicar el slice en las tuplas, obteniendo su vez nuevas tuplas con sus elementos correspondientes:

```python
# Sublista(slice) desde el índice 0 al 2

tuple2[0:3]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

Podemos obtener los últimos dos elementos de una tupla:

```python
# Sublista(slice) desde el índice 3 al 4

tuple2[3:5]

pelicula2[3:5]
```

<h3 id="nest">Tupla Anidada</h3>

Una tupla puede contener a otras tuplas, adi como otro tipo de datos complejos. Este proceso es conocido como “anidar”. Considere la siguiente tupla con algunos elementos:

```python
# Crear una tupla anidada

NestedT =(1, 2, ("pop", "rock") ,(3,4),("disco",(1,2)))

catalogo = ("El Padrino", 1972, ("Drama", "Crimen"), ("Marlon Brando", "Al Pacino"))
print(catalogo[2][0])
print(catalogo[3][1])
```

## Conclusiones

Las tuplas son útiles cuando se necesita que los datos no cambien durante la ejecución del programa, en comparación de las listas, que sí permiten modificaciones. Por otro lado, la indexación negativa facilita el acceso a los últimos elementos sin necesidad de conocer la longitud exacta de la tupla, lo que simplifica el manejo de datos de tamaño variable. Finalmente, el anidamiento permite representar estructuras de datos más complejas, aunque requiere tener claro cuántos niveles de índices hay que recorrer para llegar al valor deseado.
