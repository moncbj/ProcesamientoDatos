<h3 id="range">Rango</h3>

En ocasiones, es necesario repetir la misma operación varias veces. La repetición de operaciones es una tarea de la que se encargan los <b>bucles</b>. Veremos dos tipos, el <code>for</code> y el <code>while</code>.

Antes de discutir el tema de los bucles, hablemos primero del <code>range</code>. Es útil pensar en el rango como una lista ordenada. Por ahora, veamos un caso sencillo. Si quisiéramos generar una secuencia que contenga tres elementos ordenados desde 0 hasta 2 usaríamos este comando:

```python
# Usar el rango

range(3)
```

<h3 id="for">¿Qué es el bucle <code>for</code>?</h3>

El bucle <code>for</code> permite ejecutar un bloque de código múltiples veces. Por ejemplo, podrías usarlo si quisieras imprimir cada elemento en una lista. Usemos el bucle <code>for</code> para imprimir todos los años presentes en la lista <code>dates</code>:

Esto puede hacerse de la siguiente forma:

```python
# Ejemplo de un bucle for

dates = [1982,1980,1973]
N = len(dates)

for i in range(N):
    print(dates[i])     
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crea la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">dates</code> y se guarda su longitud en <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">N</code>. El bucle <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">for i in range(N)</code> repite el código con sangría una vez por cada valor de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">i</code> entre 0 y N-1, y en cada repetición <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">print(dates[i])</code> imprime el elemento correspondiente a esa posición.</p>
</div>

El código con sangría es ejecutado <code>N</code> veces, el valor de <code>i</code> se incrementa en 1 por cada ejecución. La sentencia para ejecutar es <code>imprimir</code> el valor de la lista de acuerdo con el índice <code>i</code> como se muestra aquí:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%203/Images/LoopsForRange.gif" width="800" />

En este ejemplo imprimiremos una secuencia de números del 0 al 7:

```python
# Ejemplo de un bucle for

for i in range(0, 8):
    print(i)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

En Python podemos acceder directamente a los elementos de la lista de la siguiente forma:

```python
# Ejemplo de un bucle for en una lista

for year in dates:  
    print(year)   
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">En vez de usar un índice, <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">for year in dates</code> recorre <b>directamente</b> los elementos de la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">dates</code>. En cada iteración, la variable <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">year</code> toma el valor de un elemento distinto, sin necesidad de usar <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">range()</code> ni corchetes.</p>
</div>

Por cada iteración, el valor de la variable <code>years</code> se comporta de forma parecida al valor de <code>dates[i]</code> del primer ejemplo:

Podemos modificar los elementos en una lista:

```python
# Usar el bucle for para cambiar los elementos de la lista

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, 5):
    print("Before square ", i, 'is',  squares[i])
    squares[i] = 'weight'
    print("After square ", i, 'is',  squares[i])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">El bucle recorre la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">squares</code> usando el índice <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">i</code>, imprime el valor original en cada posición, y luego lo <b>reemplaza</b> con <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">squares[i] = &#x27;weight&#x27;</code>. Al final del bucle, todos los elementos de la lista han sido modificados.</p>
</div>

Podemos acceder al índice y los elementos de una lista de la siguiente manera:

```python
# Bucle for en una lista haciendo iteraciones con el valor del índice y el elemento de la lista

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares):
    print(i, square)
```

<h3 id="while">¿Qué es el bucle <code>while</code>?</h3>

Como puedes observar, el bucle <code>for</code> es utilizado para controlar el flujo de la repetición. Sin embargo, ¿Qué pasa cuando no sabemos en que momento detener el bucle? ¿Y si queremos mantener la ejecución de un bloque de código hasta que se cumpla cierta condición? El bucle <code>while</code> es una herramienta para ejecutar repetidamente en base a una condición. El bloque de código seguirá ejecutándose hasta que la condición lógica dada regresa un valor **Falso**.

Supongamos que deseamos interactuar sobre la lista <code>dates</code> y detenernos en el año 1973, después imprimir el numero de iteraciones. Esto puede realizarse con el siguiente bloque de código:

```python
# Ejemplo de Bucle While

dates = [1982, 1980, 1973, 2000]

i = 0
year = 0

while(year != 1973):
    year = dates[i]
    i = i + 1
    print(year)

print("It took ", i ,"repetitions to get out of loop.")
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">El bucle <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">while(year != 1973)</code> sigue ejecutándose mientras <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">year</code> sea distinto de 1973. En cada vuelta se lee el siguiente elemento de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">dates</code> y se aumenta el contador <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">i</code>. En cuanto <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">year</code> llega a 1973, la condición se vuelve falsa y el bucle se detiene, habiendo tardado 3 repeticiones.</p>
</div>

Un bucle while interactua hasta que la condición en el argumento no es alcanzada, como se muestra en la imagen siguiente:

### Ejemplo adicional uniendo toda la teoría:
Bucles sobre un catálogo de películas

```python
# Bucle for con range()
duraciones = [175, 152, 195, 88, 136]
N = len(duraciones)

for i in range(N):
    print(duraciones[i])

# Bucle for directamente sobre la lista
peliculas = ["El Padrino", "Pulp Fiction", "Parásitos"]
for pelicula in peliculas:
    print(pelicula)

# Modificar elementos con for
calificaciones = [8.0, 7.5, 6.0, 9.0]
for i in range(len(calificaciones)):
    if calificaciones[i] >= 8:
        calificaciones[i] = "Recomendada"
print(calificaciones)

# Bucle while
i = 0
duracion_acumulada = 0
while duracion_acumulada < 300 and i < len(duraciones):
    duracion_acumulada += duraciones[i]
    i += 1
print("Películas vistas:", i, "- Minutos totales:", duracion_acumulada)
```

Escribe un bucle while que muestre los valores de la calificación de las canciones de un disco almacenados en la lista <code>PlayListRatings</code>. Si la calificación es menor a 6, sal del bucle. La lista <code>PlayListRatings</code> esta compuesta por: <code>PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]</code>

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

PlaylistRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]

i = 1
value = PlaylistRatings[0]

while(value >= 6):
    print(value)
    value = PlaylistRatings[i]
    i = i + 1
    
```

## Conclusiones

El bucle `for` es más adecuado cuando se sabe de antemano cuántas veces debe repetirse una acción, por ejemplo al recorrer todos los elementos de una lista. En cambio, el bucle `while` resulta más útil cuando el número de repeticiones depende de una condición que puede cambiar durante la ejecución del programa. Es importante tener cuidado con este último, ya que si la condición nunca se vuelve falsa, el programa entra en un ciclo infinito.
