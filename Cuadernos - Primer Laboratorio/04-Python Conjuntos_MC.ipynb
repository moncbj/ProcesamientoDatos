Comprender y aplicar las operaciones fundamentales con conjuntos en Python para el manejo de colecciones de elementos únicos.

## Objetivos Específicos

- Crear conjuntos a partir de listas y modificar su contenido utilizando los métodos `add()` y `remove()`.
- Aplicar operaciones lógicas entre conjuntos, como unión, intersección y diferencia.
- Determinar relaciones de subconjunto y superconjunto entre dos o más conjuntos.

## Resumen de Conceptos Clave

**Conjunto (set):** Colección de elementos únicos y sin orden definido. Se define con llaves `{}`, y Python elimina automáticamente los elementos duplicados.

**`add()` / `remove()`:** Permiten agregar o quitar un elemento del conjunto.

**Intersección (`&` o `intersection()`):** Devuelve los elementos que están presentes en ambos conjuntos.

**Diferencia (`difference()`):** Devuelve los elementos que están en un conjunto pero no en el otro.

**Unión (`union()`):** Devuelve todos los elementos presentes en cualquiera de los dos conjuntos.

**Subconjunto y superconjunto (`issubset()`, `issuperset()`):** Verifican si todos los elementos de un conjunto están contenidos dentro de otro.

<h3 id="content">Contenido de un Conjunto</h3>

Un conjunto es una colección de objetos únicos en Python. Un conjunto se define usando llaves de apertura y cierre <b>{}</b>. Python removerá automáticamente elementos duplicados.

```python
# Crear un conjunto

set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
set1
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crea el conjunto <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">set1</code> usando llaves <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">{}</code> con géneros musicales que incluyen repeticiones ("rock" aparece tres veces). Python elimina automáticamente los duplicados, por lo que el conjunto resultante solo conserva una copia de cada elemento.</p>
</div>

El proceso de mapeado se muestra en la siguiente figura:

También se puede crear un conjunto a partir de una lista:

```python
# Convertir una lista en un conjunto

album_list = [ "Michael Jackson", "Thriller", 1982, "00:42:19", \
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)             
album_set
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">set(album_list)</code> convierte la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">album_list</code> en un conjunto, eliminando cualquier elemento repetido y descartando el orden original, ya que los conjuntos no mantienen una secuencia definida.</p>
</div>

Vamos a crear un conjunto de géneros musicales:

```python
# Convertir una lista en un conjunto

music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul", \
                    "progressive rock", "soft rock", "R&B", "disco"])
music_genres
```

<h3 id="op">Operaciones con Conjuntos</h3>

Vamos a revisar las operaciones con conjuntos que nos ayudaran a modificarlo. Considera el conjunto <b>A</b>:

```python
# Conjunto de ejemplo

A = set(["Thriller", "Back in Black", "AC/DC"])
A
```

Este conjunto se usará en los siguientes ejemplos para practicar sus métodos.</p>

Podemos agregar un elemento al conjunto mediante el método <code>add()</code>:

```python
# Agregar un elemento al conjunto

A.add("NSYNC")
A
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

A diferencia de las listas, los elementos de un conjunto no tienen una posición fija, por lo que <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">add()</code> no permite indicar dónde insertarlo.</p>
</div>

Si agregamos el mismo elemento dos veces nada cambiara debido a que no puede haber elemento duplicados en un conjunto:

```python
# Trata de agregar un elemento duplicado al conjunto

A.add("NSYNC")
A
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se intenta agregar "NSYNC" nuevamente con <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A.add(&quot;NSYNC&quot;)</code>. Como los conjuntos no permiten elementos duplicados, el conjunto queda exactamente igual que antes; no se genera ningún error ni se agrega una segunda copia.</p>
</div>

Podemos borrar un elemento del conjunto mediante el método <code>remove</code>:

```python
# Remover un elemento del conjunto

A.remove("NSYNC")
A
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

Podemos verificar si existe un elemento especifico dentro del conjunto con el comando <code>in</code>:

Se puede averiguar si un conjunto es un súper conjunto o un sub conjunto de otro, respectivamente, de la siguiente forma:

```python
# Revisar si es un super conjunto

set(album_set1).issuperset(album_set2)   
```

Como "The Dark Side of the Moon" no está en <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">album_set1</code>, el resultado es <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">False</code>.</p>

### Ejemplo adicional uniendo toda la teoría:

```python
# Crear un conjunto de géneros (con duplicados)
generos = {"drama", "crimen", "drama", "suspenso", "crimen"}
generos

# Convertir una lista de géneros en un conjunto
lista_generos = ["drama", "acción", "acción", "comedia"]
set_generos = set(lista_generos)
set_generos

# add y remove
set_generos.add("terror")
print(set_generos)
set_generos.remove("terror")
print(set_generos)

# Operaciones lógicas entre dos catálogos
catalogo_netflix = set(["drama", "comedia", "terror"])
catalogo_hbo = set(["drama", "suspenso", "comedia"])

print("Intersección:", catalogo_netflix & catalogo_hbo)
print("Unión:", catalogo_netflix.union(catalogo_hbo))
print("Solo en Netflix:", catalogo_netflix.difference(catalogo_hbo))

# Subconjunto
generos_basicos = {"drama", "comedia"}
print(generos_basicos.issubset(catalogo_netflix))
```

Considera que la lista <code>A = [1, 2, 2, 1]</code> y el conjunto <code>B = set([1, 2, 2, 1])</code> suman <code>sum(A) = sum(B)</code>

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
A = [1,2,2,1]
B = set([1,2,2,1])

# Recuerda que los conjuntos no permiten elementos duplicados
print("Suma de A: ", sum(A))    # 6
print("Suma de B: ", sum(B))    # 3
```

## Conclusiones

Los conjuntos son útiles cuando se necesita eliminar duplicados de una colección de datos, ya que Python lo hace de forma automática al convertir una lista en un conjunto. Mientras que, las operaciones lógicas (unión, intersección, diferencia) permiten comparar dos colecciones de datos de forma directa, sin necesidad de recorrerlas manualmente mediante bucles. A diferencia de las listas y las tuplas, los conjuntos no mantienen un orden definido, por lo que no es posible acceder a sus elementos mediante índices.
