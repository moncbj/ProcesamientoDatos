## Objetivos Específicos

- Aplicar indexación, slicing y anidamiento para acceder y extraer datos dentro de una lista.

- Diferenciar entre la copia de una lista por referencia y su clonación por valor.

## Resumen de Conceptos Clave

**Lista:** Colección ordenada y mutable de elementos, que puede contener distintos tipos de datos, incluyendo otras listas o tuplas.

**Indexación y slicing:** Funcionan igual que en las tuplas, permitiendo acceder a un elemento puntual o a un rango de ellos.

**`extend()`:** Añade cada elemento de un iterable como un nuevo elemento individual de la lista.

**`append()`:** Añade el argumento completo como un único nuevo elemento al final de la lista (incluso si es una lista).

**`del`:** Elimina un elemento de la lista a partir de su índice.

**`split()`:** Convierte una cadena en una lista, dividiéndola según un separador (por defecto, espacios en blanco).

**Copia por referencia vs. clonación:** Al asignar `B = A`, ambas variables apuntan a la misma lista en memoria (si se modifica una, se modifica la otra). Al usar `B = A[:]`, se crea una copia independiente.

<h3 id="index">Indexación</h3>

Vamos a ver un poco mas sobre las listas en Python. Una lista es una colección ordenada de distintos objetos tales como enteros, cadenas e incluso otras listas. La dirección de cada elemento dentro de una lista se llama <b>índice</b>. El índice se usa para acceder o referirse a un objeto dentro de la lista.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/ListsIndex.png" width="1000" />

Para crear una lista, se escribe cada elemento dentro de <b>[ ]</b>, separado cada uno por comas. ¡Vamos a probar!

```python
# Crear una lista

L = ["Michael Jackson", 10.1, 1982]
L

peliculas = ["El Padrino", 1972, "Drama"]
peliculas
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crean dos listas usando corchetes <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">[ ]</code>: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L</code>, que mezcla texto, un decimal y un entero, y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">peliculas</code>, con datos de una película. A diferencia de la tupla, la lista puede modificarse después de creada.</p>
</div>

Podemos usar en una lista la indexación regular o negativa:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/ListsNeg.png" width="1000" />

```python
# Imprime los elementos de cada índice

print('the same element using negative and positive indexing:\n Postive:',L[0],
'\n Negative:' , L[-3]  )
print('the same element using negative and positive indexing:\n Postive:',L[1],
'\n Negative:' , L[-2]  )
print('the same element using negative and positive indexing:\n Postive:',L[2],
'\n Negative:' , L[-1]  )
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

```python
# Lista de ejemplo

["Michael Jackson", 10.1, 1982, [1, 2], ("A", 1)]
```

Esto muestra que una lista puede contener cualquier tipo de dato, incluyendo otras estructuras.</p>

Podemos usar el método <code>extend</code> para añadir elementos a la lista:

```python
# Aplica el método extend para añadir elementos

L = [ "Michael Jackson", 10.2]
L.extend(['pop', 10])
L

peliculas_a = ["El Padrino", 1972]
peliculas_a.extend(["Drama", 9.2])
peliculas_a
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L.extend([&#x27;pop&#x27;, 10])</code> añade cada elemento del argumento como un elemento individual y nuevo al final de la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L</code>, que queda con cuatro elementos en total.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Lo mismo ocurre con <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">peliculas_a.extend([&quot;Drama&quot;, 9.2])</code>: se agregan "Drama" y 9.2 como dos elementos separados al final de la lista.</p>
</div>

Otro método similar es <code>append</code>. Si usamos <code>append</code> en vez de <code>extend</code>, añadimos un elemento a a la lista:

```python
# Aplica el método append para añadir elementos

L = [ "Michael Jackson", 10.2]
L.append(['pop', 10])
L

peliculas_b = ["El Padrino", 1972]
peliculas_b.append(["Drama", 9.2])
peliculas_b
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">A diferencia de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">extend()</code>, el método <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">append()</code> agrega el argumento completo como <b>un solo</b> elemento nuevo. Por eso <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L.append([&#x27;pop&#x27;, 10])</code> añade la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">[&#x27;pop&#x27;, 10]</code> como un único elemento anidado, en vez de dos elementos separados.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Igual sucede con <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">peliculas_b.append([&quot;Drama&quot;, 9.2])</code>: el resultado queda con la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">[&quot;Drama&quot;, 9.2]</code> anidada dentro de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">peliculas_b</code>.</p>
</div>

Cada vez que un método es aplicado, la lista cambia. Al usar <code>extend</code> estamos añadiendo dos nuevos elementos. La lista <code>L</code> es entonces modificada con dos nuevos elementos.

```python
# Aplica el método extend para añadir elementos

L = [ "Michael Jackson", 10.2]
L.extend(['pop', 10])
L
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se repite el ejemplo de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">extend()</code> desde cero, partiendo de una lista con dos elementos y agregando "pop" y 10 como dos nuevos elementos individuales.</p>
</div>

Si añadimos a la lista <code>['a','b']</code> tendremos un nuevo elemento conformado por una lista anidada:

```python
# Aplica el método append para añadir elementos

L.append(['a','b'])
L
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L.append([&#x27;a&#x27;,&#x27;b&#x27;])</code> añade la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">[&#x27;a&#x27;, &#x27;b&#x27;]</code> como un único elemento nuevo al final de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">L</code>, quedando anidada dentro de la lista principal en vez de agregar "a" y "b" por separado.</p>
</div>

Las listas son estructuras que pueden cambiar con el tiempo. Por ejemplo, cambiemos el primer elemento de una lista de la siguiente manera:

```python
# Cambia el elemento usando el índice

A = ["disco", 10, 1.2]
print('Before change:', A)
A[0] = 'hard rock'
print('After change:', A)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Las listas son mutables: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A[0] = &#x27;hard rock&#x27;</code> reemplaza el elemento en la posición 0 (que era "disco") por el nuevo valor "hard rock", modificando la lista original directamente.</p>
</div>

Para borrar un elemento de la lista, usamos el comando <code>del</code>:

```python
# Borra el elemento usando el índice

print('Before change:', A)
del(A[0])
print('After change:', A)

peliculas_a[0] = "El Padrino II"
print(peliculas_a)
del(peliculas_a[-1])
print(peliculas_a)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">del(A[0])</code> elimina el elemento en la posición 0 de la lista, desplazando el resto de los elementos una posición hacia adelante.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">De igual forma, <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">peliculas_a[0] = &quot;El Padrino II&quot;</code> modifica el primer elemento, y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">del(peliculas_a[-1])</code> elimina el último elemento de la lista usando índice negativo.</p>
</div>

Podemos transformar una cadena en una lista con <code>split</code>. Por ejemplo, el método <code>split</code> convierte cada grupo de caracteres separados por un espacio en blanco, en un elemento de la lista:

```python
# Divide la cadena

'hard rock'.split()

generos = "Drama,Crimen,Suspenso".split(",")
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">&#x27;hard rock&#x27;.split()</code> convierte la cadena en una lista, dividiéndola en cada espacio en blanco; el resultado es <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">[&#x27;hard&#x27;, &#x27;rock&#x27;]</code>.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">De forma similar, <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">&quot;Drama,Crimen,Suspenso&quot;.split(&quot;,&quot;)</code> divide la cadena usando la coma como separador, generando la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">generos</code> con tres elementos.</p>
</div>

Podemos utilizar la función split para separar distintas cadenas usando algún carácter especifico. Pasamos en el argumento el carácter que queremos usar como separador, en este caso será una coma. El resultado es una lista, y cada elemento corresponderá a un grupo de caracteres separados por una coma:

```python
# Separa la cadena usando comas

'A,B,C,D'.split(',')
```

<h3 id="co">Copiado y Clonación de Listas</h3>

Cuando asignamos el valor de la variable <b>B</b> igual a <b>A</b>; ambas están haciendo referencia a la misma lista guardada en memoria:

```python
# Copiar (por referencia) la lista A

A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Al escribir <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B = A</code> no se crea una lista nueva: ambas variables quedan apuntando a la <b>misma</b> lista en memoria. Por eso <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">print(&#x27;A:&#x27;, A)</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">print(&#x27;B:&#x27;, B)</code> muestran el mismo contenido.</p>
</div>

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/ListsRef.png" width="1000" align="center">

En un principio, el valor del primer elemento en <b>B</b> es asignado como hard rock. Si modificamos el primer elemento en <b>A</b> por <b>banana</b>, obtendremos un efecto inesperado. Debido a que <b>A</b> y <b>B</b> hacen referencia a la mima lista, si cambiamos <b>A</b>, entonces cambiamos también a <b>B</b>. Al revisar el primer elemento de <b>B</b> veremos banana en vez de hard rock:

```python
# Examinar la copia hecha por referencia

print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Como <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code> apuntan a la misma lista, al modificar <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A[0] = &quot;banana&quot;</code> el cambio también se refleja en <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code>: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B[0]</code> pasa de "hard rock" a "banana" aunque nunca se modificó <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code> directamente.</p>
</div>

Se muestra esto en la siguiente imagen:

<img src = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/ListsRefGif.gif" width="1000" />

Se puede clonar la lista **A** haciendo uso de la siguiente sintaxis:

```python
# Clonar la lista A

B = A[:]
B
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B = A[:]</code> usa slicing sin límites para crear una <b>copia independiente</b> (clon) de la lista <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A</code>, en lugar de una referencia a la misma lista.</p>
</div>

La variable **B** es una copia o clon por valor de la lista original; esto se puede demostrar en la siguiente imagen:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/ListsVal.gif" width="1000" />

Ahora cualquier cambio en <b>A<b/>, no será aplicado a <b>B</b>:

```python
print('B[0]:', B[0])
A[0] = "hard rock"
print('B[0]:', B[0])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Como ahora <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code> es un clon independiente de <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A</code>, modificar <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A[0] = &quot;hard rock&quot;</code> no afecta a <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code>: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B[0]</code> conserva su valor original.</p>
</div>

```python
# Ejemplo de copia de referencia vs clonación
A = ["Drama", "Crimen"]
B = A          # referencia
C = A[:]       # clon
A[0] = "Comedia"
print("B (referencia):", B)
print("C (clon):", C)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Este ejemplo compara directamente ambos comportamientos: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B = A</code> crea una referencia (comparten la misma lista), mientras que <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">C = A[:]</code> crea un clon independiente.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Al modificar <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A[0] = &quot;Comedia&quot;</code>, <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">B</code> cambia junto con <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A</code> (porque es la misma lista), pero <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">C</code> conserva su valor original, ya que es una copia separada.</p>
</div>

### Ejemplo adicional uniendo toda la teoría:
Tupla con información de una película

```python
# Crear una lista de películas
peliculas = ["El Padrino", 1972, "Drama"]
peliculas

# Slicing
print(peliculas[1:3])

# extend vs append
peliculas_a = ["El Padrino", 1972]
peliculas_a.extend(["Drama", 9.2])
print(peliculas_a)

peliculas_b = ["El Padrino", 1972]
peliculas_b.append(["Drama", 9.2])
print(peliculas_b)

# Modificar y borrar elementos
peliculas_a[0] = "El Padrino II"
print(peliculas_a)
del(peliculas_a[-1])
print(peliculas_a)

# split
generos = "Drama,Crimen,Suspenso".split(",")
print(generos)

# Copia por referencia vs clonación
A = ["Drama", "Crimen"]
B = A          # referencia
C = A[:]       # clon
A[0] = "Comedia"
print("B (referencia):", B)
print("C (clon):", C)
```

## Conclusiones

La diferencia entre `extend()` y `append()` es clave: `extend()` agrega elementos individuales a la lista, mientras que `append()` agrega el objeto completo, incluso si este es otra lista. Por otro lado, entender la diferencia entre copiar por referencia y clonar una lista resulta importante para evitar modificar datos sin querer, ya que ambas variables pueden estar apuntando al mismo objeto en memoria. Como resultado, las listas resultan más flexibles que las tuplas, pues se pueden modificar después de creadas, lo que las hace apropiadas cuando los datos van a cambiar a lo largo del programa.
