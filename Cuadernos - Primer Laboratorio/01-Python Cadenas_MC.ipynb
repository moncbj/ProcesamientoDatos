<h2 id="strings">¿Qué es una cadena?</h2>

A continuación, se muestra una cadena entre comillas dobles.

```python
# Usa las comillas dobles para definir una cadena

"Michael Jackson"

"El Padrino"
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se escriben dos cadenas usando comillas dobles: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">"Michael Jackson"</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">"El Padrino"</code>.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Como no se usa <code>print()</code> ni se guardan en una variable, Jupyter solo muestra en pantalla el resultado de la <b>última línea</b> de la celda (por eso solo se ve "El Padrino" como salida, aunque las dos cadenas sí se crearon).</p>

También podemos usar comillas simples:

```python
# Usa comillas simples para definir una cadena

'Michael Jackson'

'El Padrino'
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

En Python ambas son equivalentes para definir una cadena; se elige una u otra por preferencia o para poder incluir el otro tipo de comilla dentro del texto.</p>
</div>

Una cadena puede ser una combinación de espacios y dígitos:

```python
# Dígitos y espacios en una cadena

'1 2 3 4 5 6 '

'- d 4 . '
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crean dos cadenas que combinan dígitos, espacios y puntos: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">'1 2 3 4 5 6 '</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">'- d 4 . '</code>. Aunque contienen números, Python las trata como texto (tipo <code>str</code>) porque están entre comillas, no como valores numéricos.</p>
</div>

Una cadena también puede estar formada por caracteres especiales :

```python
# Caracteres especiales en una cadena

'@#2_#]&*^%$'
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se define la cadena <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">'@#2_#]&*^%$'</code>, formada solo por símbolos. Esto demuestra que una cadena en Python puede contener cualquier carácter disponible en el teclado, no solo letras.</p>
</div>

Podemos imprimir en pantallas una cadena usando la sentencia print:

```python
# Imprimir la cadea

print("hello!")
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

A diferencia de escribir la cadena sola, <code>print()</code> imprime el contenido sin las comillas alrededor.</p>

Podemos asignar una cadena como valor de una variable:

```python
# Asignar una cadena a una variable

Name = "Michael Jackson"
Name

titulo = "El Padrino"
titulo
```

Desde este punto, cada vez que se use <code>Name</code> o <code>titulo</code> en el cuaderno, Python usa el texto que se les asignó aquí.</p>
</div>

<h2 id="index">Indexación</h2>

Debemos pensar en una cadena como una secuencia ordenada. Cada elemento de la secuencia puede ser accedido utilizando un índice numérico, el cual representa el lugar que ocupa dicho elemento en la cadena:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%201/Images/StringsIndex.png" width="600" align="center" />

Se puede acceder al primer índice de la siguiente manera:

<hr/>
<div class="alert alert-success alertsuccess" style="margin-top: 20px">
[Tip]: En la indexación se empieza por el 0, esto quiere decir que el primer elemento de la cadena está en el índice 0.
</div>
<hr/>

```python
# Imprime el primer elemento de la cadena

print(Name[0])

print(titulo[0])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[0]</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[0]</code> acceden al carácter que está en la posición 0 de cada cadena.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">En Python la indexación empieza en 0, no en 1, por lo que <code>[0]</code> siempre corresponde al primer carácter de la cadena.</p>
</div>

Podemos acceder el índice 6:

```python
# Imprime el elemento en el índice 6 de la cadena

print(Name[6])

print(titulo[6])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[6]</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[6]</code> acceden al carácter en la posición 6 (es decir, el séptimo carácter, ya que se cuenta desde 0).</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">En <code>Name = "Michael Jackson"</code>, la posición 6 corresponde a la letra "J" de "Jackson" (contando el espacio como una posición más).</p>
</div>

También, podemos acceder al treceavo elemento:

```python
# Imprime el elemento en el treceavo índice de la cadena

print(Name[13])

# print(titulo[13]) En este caso el indice no existe, saldrá error
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>

Si se ejecutara sin el comentario, lanzaría un error <code>IndexError</code> por pedir una posición que no existe.</p>
</div>

<h3 id="neg">Indexación Negativa</h3>

Con las cadenas tambien podemos usar la indexación negativa:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%201/Images/StringsNeg.png" width="600" align="center" />

La indexación negativa nos sirve para numerar un elemento desde el final de una cadena.

El ultima elemento tiene el índice -1:

```python
# Imprime el último elemento de la cadena

print(Name[-1])

print(titulo[-1])
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[-1]</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[-1]</code> usan indexación negativa para acceder al <b>último carácter</b> de cada cadena.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">El índice -1 siempre apunta al último elemento, sin importar cuántos caracteres tenga la cadena; por eso es más práctico que calcular la longitud para llegar al final.</p>
</div>

El primer elemento se obtiene usando como índice el -15:

```python
# Imprime el primer elemento de la cadena

print(Name[-15])

len(titulo)
print(titulo[-10])  # En este caso el total de la cadena es 10, por lo que -10 es el primer elemento
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[-15]</code> accede al carácter 15 posiciones antes del final. Como <code>Name</code> tiene 16 caracteres, la posición -15 corresponde al primer carácter (índice 0).</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">len(titulo)</code> calcula la longitud de <code>titulo</code>, que es 10. Por eso <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[-10]</code> también corresponde al primer carácter: el índice negativo que llega hasta el primero siempre es <code>-len(cadena)</code>.</p>
</div>

Para conocer la cantidad de caracteres en un cadena usamos <code>len</code>, que es la abreviación de length, en inglés significa longitud:

```python
# Encuentra la longitud de la cadena

len("Michael Jackson")
len(titulo)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">len(&quot;Michael Jackson&quot;)</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">len(titulo)</code> devuelven la cantidad de caracteres de cada cadena (incluyendo espacios).</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">"Michael Jackson" tiene 15 caracteres y "El Padrino" tiene 10.</p>
</div>

<h3 id="slice">Slicing</h3>

Podemos obtener múltiples caracteres de una cadena usando el slicing, digamos desde primero hasta el cuarto y desde el octavo hasta el doceavo:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%201/Images/StringsSlice.png" width="600" align="center" />

<hr/>
<div class="alert alert-success alertsuccess" style="margin-top: 20px">
[Tip]: Para aplicar el slicing, el primer numero indica el índice del primer elemento, y el segundo numero indicara el índice del ultimo elemento de la sub-cadena que queremos obtener.
</div>
<hr/>

```python
# Asigna una seccion de la cadena dentro de la variable Name usando los indices desde el 0 hasta el 3

Name[0:4]
titulo[0:3]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[0:4]</code> aplica slicing para extraer los caracteres desde el índice 0 hasta el 3 (el índice final, 4, no se incluye). El resultado es "Mich".</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[0:3]</code> extrae los caracteres en las posiciones 0, 1 y 2, es decir "El " (incluyendo el espacio).</p>
</div>

```python
# Asigna una seccion de la cadena dentro de la variable Name usando los indices desde el 8 hasta el 11

Name[8:12]
titulo[8:12]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[8:12]</code> extrae los caracteres desde la posición 8 hasta la 11 (sin incluir la 12), lo que corresponde a "ckso" dentro de "Michael Jackson".</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">titulo[8:12]</code> intenta extraer desde la posición 8 hasta la 11, pero como <code>titulo</code> solo llega hasta el índice 9, Python simplemente devuelve hasta donde alcance la cadena, sin lanzar error.</p>
</div>

<h3 id="stride">Stride</h3>

Para usar la técnica del stride debemos asignar un valor que a continuación se explica, si usamos el '2' por ejemplo, estaremos indicando que queremos seleccionar solamente cada segundo elemento de la cadena:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%201/Images/StringsStride.png" width="600" align="center" />

```python
# Toma cada segundo elemento. Los elementos con índice 1, 3, 5 ...

Name[::2]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[::2]</code> usa slicing con paso (stride). Al omitir el inicio y el fin, Python recorre toda la cadena, pero el <code>2</code> después del segundo <code>:</code> indica que tome un carácter sí y uno no (posiciones 0, 2, 4, 6...).</p>
</div>

También Podemos combinar el slicing con el stride. Para este caso, seleccionamos los primeros cinco elementos y después aplicamos stride:

```python
# Toma cada segundo elemento desde el índice 0 al 4

Name[0:5:2]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name[0:5:2]</code> combina slicing y stride: primero delimita el rango entre las posiciones 0 y 4 (sin incluir la 5), y dentro de ese rango toma cada segundo carácter (posiciones 0, 2 y 4).</p>
</div>

<h3 id="concat">Concatenar Cadenas</h3>

Podemos concatenar o combinar cadenas usando el símbolo de suma, el resultado será una nueva cadena que surge de la combinación de ambas:

```python
# Concatenar dos cadenas

Statement = Name + "is the best"
Statement

ficha = titulo + " (1972)"
ficha
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Statement = Name + &quot;is the best&quot;</code> y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">ficha = titulo + &quot; (1972)&quot;</code> usan el operador <code>+</code> para concatenar (unir) dos cadenas en una nueva.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">La concatenación no modifica las cadenas originales; crea una cadena completamente nueva y la guarda en la variable de la izquierda.</p>
</div>

Para replicar los valores en una cadena simplemente podemos multiplicarla por el numero de veces que deseemos. Para este caso, el numero de veces es tres. El resultado es una nueva cadena, la cual consiste en tres copias de la original:

```python
# Imprime la cadena 3 veces

3 * "Michael Jackson"
```

<h2 id="escape">Secuencias de Escape</h2>

Una barra invertida representa el inicio de una secuencia de escape. Estas representan cadenas que pudieran ser difíciles de introducir. Por ejemplo, una barra invertida seguida de una "n" representa un salto de línea. En la salida de pantalla tendremos un salto de línea cada vez que una barra invertida y la letra "n" sea encontrada.

```python
# Secuencia de escape para un Salto de línea

print(" Michael Jackson \n is the best" )

sinopsis = "Un patriarca cede el control de su imperio criminal.\nGénero: Drama"
print(sinopsis)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">\n</code> dentro de una cadena es una secuencia de escape que representa un salto de línea. Por eso al imprimir el texto, este se corta y continúa en la línea siguiente.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Lo mismo ocurre en <code>sinopsis</code>: el texto se corta en el punto donde aparece <code>\n</code>, mostrando "Género: Drama" en una línea aparte.</p>
</div>

De igual forma, barra invertida "t" aplicará un espacio de tabulación:

```python
# Secuencia de escape para un espacio de tabulación

print(" Michael Jackson \t is the best" )
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">\t</code> es la secuencia de escape para un tabulador. Al imprimir la cadena, Python reemplaza <code>\t</code> por un espacio de tabulación, separando visualmente las dos partes del texto.</p>
</div>

En caso de quieras que una barra invertida forme parte de una cadena, utiliza una doble barra invertida:

```python
# Incluir una barra invertida en una cadena

print(" Michael Jackson \\ is the best" )
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Escribir doble barra invertida le indica a Python que se quiere imprimir una barra invertida literal, en lugar de que la interprete como el inicio de una secuencia de escape.</p>
</div>

Otra forma de hacer que se muestre la barra invertida de una cadena es colocando antes una “r”:

```python
# La letra r le dirá a python que muestre la cadena tal cual e ignore lo que haya despues de la barra invertida

print(r" Michael Jackson \ is the best" )
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Anteponer la letra <code>r</code> antes de las comillas convierte la cadena en un "raw string" (cadena cruda). Esto le indica a Python que <b>no interprete</b> ninguna secuencia de escape, y muestre la barra invertida tal como se escribió.</p>
</div>

<h2 id="operations">Operaciones con Cadenas</h2>

En Python existen muchos métodos para hacer operaciones con cadenas que pueden ser usados para manipular datos. Usaremos algunas operaciones básicas con cadenas.

Usemos ahora el método <code>upper</code>; este método convierte los caracteres de minúsculas a mayúsculas:

```python
# Convertir todos los caracteres de la cadena en mayúsculas

A = "Thriller is the sixth studio album"
print("before upper:", A)
B = A.upper()
print("After upper:", B)
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A = &quot;Thriller...&quot;</code> guarda el texto original en <code>A</code>. El método <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A.upper()</code> devuelve una <b>nueva</b> cadena con todos los caracteres en mayúscula, la cual se guarda en <code>B</code>.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">El método <code>upper()</code> no modifica <code>A</code>; por eso al imprimir "before upper" y "After upper" se ve que <code>A</code> sigue igual y <code>B</code> es la versión nueva en mayúsculas.</p>
</div>

El método <code>replace</code> reemplaza un segmento de la cadena, p.ej. una sub-cadena por una nueva. Introducimos primero la parte de la cadena que queremos cambiar. En el segundo argumento pondremos el segmento que reemplazara al anterior, el resultado será una nueva cadena con una de sus partes modificada:

```python
# Reemplazar la vieja sub-cadena por una nueva

A = "Michael Jackson is the best"
B = A.replace('Michael', 'Janet')
B
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;"> Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">A.replace(&quot;Michael&quot;, &quot;Janet&quot;)</code> busca todas las apariciones de "Michael" dentro de <code>A</code> y las reemplaza por "Janet", devolviendo una cadena nueva guardada en <code>B</code>. La cadena original <code>A</code> no se modifica, porque las cadenas en Python son inmutables.</p>
</div>

La función del método <code>find</code> es encontrar una sub-cadena. El argumento del método es la sub-cadena que deseas encontrar, y el resultado será el numero del índice perteneciente al primer carácter de la sub-cadena. Podemos encontrar la sub-cadena <code>jack</code> o <code>el</code>.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%201/Images/StringsFind.png" width="600" align="center" />

```python
# Encontrar la sub-cadena. El resultado será unicamente el índice del primer elemente de la sub-cadena dentro de la cadena principal.

Name = "Michael Jackson"
Name.find('el')
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Name.find(&quot;el&quot;)</code> busca la subcadena "el" dentro de <code>Name</code> y devuelve el índice donde comienza la <b>primera</b> aparición. <code>find()</code> siempre reporta la primera coincidencia, aunque el texto se repita más adelante en la cadena.</p>
</div>

```python
# Encotrar la sub-cadena dentro de la cadena.

Name.find('Jack')
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">

</div>

Si la sub-cadena no se encuentra en la cadena principal el resultado será -1. Por ejemplo, la cadena 'Jasdfasdasdf' no es una sub-cadena:

```python
# Si no se encuentra la sub-cadena en la cadena principal

Name.find('Jasdfasdasdf')
```

### Ejemplo adicional uniendo toda la teoría:
Cadenas con datos de películas

```python
# Cadena de ejemplo: título de una película
titulo = "El Padrino"

# Indexación positiva y negativa
print(titulo[0])
print(titulo[-1])

# Slicing: obtener las primeras 3 letras
print(titulo[0:3])

# Stride: obtener cada segundo carácter
print(titulo[::2])

# Concatenación
ficha = titulo + " (1972)"
print(ficha)

# Secuencia de escape
sinopsis = "Un patriarca cede el control de su imperio criminal.\nGénero: Drama"
print(sinopsis)

# Métodos de cadena
print(titulo.upper())
print(titulo.replace("Padrino", "Padrino II"))
print(titulo.find("Padrino"))
print(len(titulo))
```

<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Este bloque reúne, en un solo ejemplo, todas las operaciones vistas en el cuaderno pero aplicadas a un dato distinto (el título de una película en vez de un nombre):</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Indexación</b> (<code>titulo[0]</code>, <code>titulo[-1]</code>) para tomar el primer y el último carácter.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Slicing</b> (<code>titulo[0:3]</code>) para extraer las primeras tres letras.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Stride</b> (<code>titulo[::2]</code>) para tomar un carácter de cada dos.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Concatenación</b> (<code>titulo + " (1972)"</code>) para armar una ficha con el año.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Secuencia de escape</b> (<code>\n</code>) para separar la sinopsis en dos líneas.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><b>Métodos de cadena</b> (<code>upper()</code>, <code>replace()</code>, <code>find()</code>, <code>len()</code>) para transformar y consultar el texto.</p>

## Conclusiones

La indexación y el slicing son la base para manipular cadenas texto en Python de forma precisa, mientras la indexación permite aislar un carácter puntual, el slicing y el stride permiten trabajar con fragmentos completos o patrones regulares dentro de una cadena, lo que resulta esencial al procesar texto de longitud variable sin conocer su tamaño de antemano. El uso de índices negativos resulta una alternativa más práctica que calcular manualmente la longitud de la cadena con `len()`, ya que se adapta automáticamente a cadenas de distintos tamaños sin necesidad de ajustar el código. Los métodos incorporados del tipo `str` (`upper()`, `replace()`, `find()`) evidencian que Python ofrece herramientas nativas suficientes para tareas comunes de procesamiento de texto (normalización, sustitución y búsqueda) sin depender de librerías externas, lo cual facilita su uso en scripts simples o etapas iniciales de análisis de datos.
