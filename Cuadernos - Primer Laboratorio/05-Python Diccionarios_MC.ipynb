En este cuaderno se trabaja con los diccionarios en Python, una estructura de datos que almacena información en pares de llave y valor. Se abordan la creación de diccionarios, el acceso a valores mediante llaves, y las operaciones de agregar, eliminar y verificar la existencia de elementos.

## Objetivos Específicos

- Modificar un diccionario agregando y eliminando pares de llave-valor.
- Utilizar los métodos `keys()` y `values()` para obtener todas las llaves y valores de un diccionario.

## Resumen de Conceptos Clave

**Diccionario (dict):** Colección de pares llave-valor. Las llaves deben ser de un tipo inmutable (cadena, número o tupla), mientras que los valores pueden ser de cualquier tipo.

**Acceso por llave:** Se obtiene el valor asociado a una llave escribiendo `diccionario['llave']`.

**`keys()` / `values()`:** Devuelven todas las llaves o todos los valores almacenados en el diccionario.

**Agregar y eliminar:** Se agrega un par nuevo asignando un valor a una llave (`dict['nueva_llave'] = valor`), y se elimina con `del(dict['llave'])`.

**Verificación de existencia (`in`):** Permite comprobar si una llave está presente en el diccionario.

<h2 id="Dic">Diccionarios</h2>

<h3 id="content">¿Qué son los diccionarios?</h3>

Un diccionario esta conformado por llaves y valores. Un diccionario puede compararse a una lista. En vez de índices numéricos como los de las listas, los diccionarios tienen llaves. Estas llaves son usadas para acceder a los valores dentro de los diccionarios.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsList.png" width="650" />

Un ejemplo de Diccionario <code>Dict</code>:

```python
# Crear el diccionario

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
Dict
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crea el diccionario <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Dict</code> con pares llave-valor separados por dos puntos <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">:</code>. Las llaves pueden ser cadenas (<code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">&quot;key1&quot;</code>), o incluso tuplas (<code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">(0, 1)</code>), y los valores pueden ser de cualquier tipo: enteros, cadenas, listas o tuplas.</p>
</div>

Las llaves pueden ser cadenas:

```python
# Acceder a un valor mediante una llave

Dict["key1"]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Dict[&quot;key1&quot;]</code> accede al valor asociado a la llave "key1", devolviendo el número 1.</p>
</div>

Una llave puede ser cualquier tipo de objeto inmutable, pudiendo ser una tupla:

```python
# Acceder a un valor mediante una llave

Dict[(0, 1)]
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;"><code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">Dict[(0, 1)]</code> muestra que una tupla también puede usarse como llave, ya que es un tipo de dato inmutable; el valor asociado a esa llave es 6.</p>
</div>

Cada llave esta separada de su respectivo valor mediante dos puntos "<code>:</code>". Las comas separan los elementos y todo el diccionario esta encerrado usando llaves. Un diccionario vacío se representa escribiendo solo dos llaves, de esta forma "<code>{}</code>".

```python
# Crear un ejemplo de diccionario

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Se crea el diccionario <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">release_year_dict</code>, donde cada llave es el nombre de un disco y su valor es el año de estreno (como cadena). La barra invertida <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">\\</code> al final de cada línea permite continuar la instrucción en la siguiente sin afectar el contenido del diccionario.</p>
</div>

Resumiendo, al igual que una lista, un diccionario guarda una secuencia de elementos. Cada uno es representado por una llave y su valor correspondiente. Los diccionarios se crean mediante dos llaves y estas contendrán las llaves y valores separados por una coma. A cada llave solo puede corresponderle un valor, sin embargo, múltiples llaves pueden tener el mismo. Las llaves solo pueden ser cadenas, números o tuplas, pero los valores pueden ser de cualquier tipo.

Es de ayuda visualizar un diccionario como una tabla, como se muestra en la imagen. La primera columna representa las llaves, la segunda los valores.

### Ejemplo adicional uniendo toda la teoría:
Diccionario de péliculas y años de estreno

```python
# Crear un diccionario de películas y su año de estreno
year_dict = {"El Padrino": 1972, "Pulp Fiction": 1994, 
             "Parásitos": 2019, "Coco": 2017}
year_dict

# Acceder a un valor mediante su llave
print(year_dict["Pulp Fiction"])

# keys() y values()
print(year_dict.keys())
print(year_dict.values())

# Agregar una entrada
year_dict["Oppenheimer"] = 2023
print(year_dict)

# Eliminar una entrada
del(year_dict["Coco"])
print(year_dict)

# Verificar existencia
print("Parásitos" in year_dict)
```

## Conclusiones

Los diccionarios son útiles cuando se necesita relacionar dos datos entre sí, como un nombre y su valor correspondiente, en lugar de únicamente almacenar una lista de datos sueltos. Por otro lado, acceder a los valores mediante llaves resulta más claro y directo que hacerlo por índice numérico, como ocurre en las listas o tuplas. Es importante tener en cuenta que las llaves de un diccionario deben ser de un tipo inmutable, razón por la cual no es posible utilizar una lista como llave.
