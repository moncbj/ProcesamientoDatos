#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a,b):
    pass
    #Colocar aqui la respuesta
    if(a>b):
      selected_num = b
      no_selected = a
    else:
      selected_num = a
      no_selected = b
    if a%2==0 and b%2==0:
      return selected_num
    else:
      return no_selected
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
  pass
  text = text.split()
  if text[0][0] == text[1][0]:
    return True
  else:
    return False
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1,n2):
  pass
  if n1+n2 == 20 or n1 == 20 or n2 == 20:
    return True
  else:
    return False
```

#### MAYÚSCULAS: Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto
     
    old_macdonald('macdonald') --> MacDonald
    
Note: `'macdonald'.capitalize()` returns `'Macdonald'`

```python
def old_macdonald(name):
    pass
    if len(name) > 3:
        return name[:3].capitalize() + name[3:].capitalize()
    else:
        return 'El nombre no es lo suficientemente largo'
```

Si el nombre tiene más de 3 caracteres, se divide en dos partes con slicing: <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">name[:3]</code> (los primeros tres caracteres) y <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">name[3:]</code> (el resto). A cada parte se le aplica <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">capitalize()</code>, que pone en mayúscula solo su primera letra, y luego se concatenan.</p><p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">Como resultado, se ponen en mayúscula tanto la primera letra del nombre completo como la primera letra del cuarto carácter en adelante, simulando "MacDonald" a partir de "macdonald".</p></div>

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    pass
    text = text.split()
    text.reverse()
    result = " "
    return result.join(text)
```

#### PROBLEMA 33: 

Dada una lista de enteros, retornar True si el arreglo tinene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python

def has_33(nums):
    pass
    for i in range(0, len(nums)-1):
        if nums[i] == 3 and nums[i+1] == 3:
            return True
    return False
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):
    pass
    result = ''
    for char in text:
        result += char * 3
    return result
```

#### BLACKJACK: Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 *y* hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'
    blackjack(5,6,7) --> 18
    blackjack(9,9,9) --> 'BUST'
    blackjack(9,9,11) --> 19

```python
def blackjack(a,b,c):
    if sum((a,b,c)) <= 21:
        return sum((a,b,c))
    elif sum((a,b,c)) <= 31 and 11 in (a,b,c):
        return sum((a,b,c)) - 10
    else:
        return 'BUST'
```

<div style="background:#f7f7f8; border-left:4px solid #7a7a7a; border-radius:4px; padding:12px 18px; font-family:Arial,sans-serif;">
<p style="margin:0 0 8px 0; font-size:13.5px; font-weight:700; color:#2c2c2c;">Explicación del bloque</p>
<p style="margin:6px 0; font-size:14px; line-height:1.55; color:#2c2c2c;">La función <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">blackjack(a, b, c)</code> se completó agregando dos casos que faltaban en el enunciado original: si la suma de los tres números supera 21 <b>y</b> alguno de ellos es un 11 (el valor "blando" del As), se resta 10 a la suma (<code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">sum((a,b,c)) - 10</code>) antes de devolverla; y si aun así la suma sigue superando 21, se devuelve la cadena <code style="background:#eef0f2; color:#333; padding:1px 6px; border-radius:4px; font-size:13px;">&#x27;BUST&#x27;</code> en lugar de un número.</p>
</div>

```python
# Revisión
blackjack(5,6,7)
```
